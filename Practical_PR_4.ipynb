# **Libraries and Import the Dataset**

```python
import numpy as np
import pandas as pd


Data = {
    "Mathematics" : [80 , 65 , 90 , 55 , 75] ,
    "Statistics" : [75 , 70 , 85 , 60 , 80] ,
    "Python" : [90 , 60 , 95 , 65 , 70]
    }


DF = pd.DataFrame(Data , index = ["S1" , "S2" , "S3" , "S4" , "S5"])
```

```python
X = DF.to_numpy()
print(X)
```

# **Part A : Vector & Matrix Fundamentals**

# **1. Represent each student’s subject scores as a vector.**

```python
S1 = X[0]
S2 = X[1]
S3 = X[2]
S4 = X[3]
S5 = X[4]

print(f"S1 : {S1}")
print(f"S2 : {S2}")
print(f"S3 : {S3}")
print(f"S4 : {S4}")
print(f"S5 : {S5}")
```

# **2. Compute : (1) Norm-1 and Norm-2 of vectors.**

```python
L1 = np.linalg.norm(S1 , ord = 1)
L2 = np.linalg.norm(S1 , ord = 2)

print(f"L1 Norm : {L1}")
print(f"L2 Norm : {L2}")
```

# **2. Compute : (2) Dot product and angle between two students’ score vectors.**

```python
Dot = np.dot(S1 , S2)

print(f"Dot Product : {Dot}")
```

# **2. Compute : (3) Cross product (for 3D selected subjects).**

```python
Cross = np.cross(S1 , S2)

print(f"Cross Product : {Cross}")
```

# **3. Find the projection of one vector onto another.**

```python
Projection = (np.dot(S1 , S2) / np.dot(S2 , S2)) * S2

print(f"Projection : {Projection}")
```

# **Part B : Matrix Operations**

# **4. Form a matrix of students × subjects. Perform : (1) Matrix addition and multiplication.**

```python
# Addition

B = np.ones((5, 3)) * 5

Addition = X + B

print(f"Matrix Addition : \n {Addition}")


# ----------------------------------------------------------------------------------------------------------------------------------------------------


# Multiplication

Multiplication = X.T @ X

print(f"\n\nMatrix Multiplication : \n {Multiplication}")
```

# **4. Form a matrix of students × subjects. Perform : (2) Transpose and Inverse (if possible).**

```python
# Transpose

Transpose = X.T

print(f"Transpose : \n {Transpose}")


# ----------------------------------------------------------------------------------------------------------------------------------------------------


# Inverse

Inverse = np.linalg.inv(Multiplication)

print(f"\n\nInverse : \n {Inverse}")
```

# **4. Form a matrix of students × subjects. Perform : (3) Determinant.**

```python
Determinant = np.linalg.det(X.T @ X)

print(f"Determinant : {Determinant}")
```

# **Part D : Eigenvalues & Decomposition**

# **7. Compute the eigenvalues and eigenvectors of the covariance matrix.**

```python
Covariance = np.cov(X , rowvar = False)

Eigenvalues = np.linalg.eigh(Covariance)[0]
Eigenvectors = np.linalg.eigh(Covariance)[1]

First = np.argsort(Eigenvalues)[::-1]

Eigenvalues = Eigenvalues[First]
Eigenvectors = Eigenvectors[ : , First]

print(f"Eigenvalues : {Eigenvalues}")
print(f"Eigenvectors : {Eigenvectors}")
```
